# Market Data Visualizations

Сравнение рыночных данных Binance и СПБ Биржи для BTC и ETH.

```python
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
```

```python
btc = pd.read_csv(
    "aggregated/btc_aggregated_1s.csv",
    parse_dates=["timestamp"]
)

eth = pd.read_csv(
    "aggregated/eth_aggregated_1s.csv",
    parse_dates=["timestamp"]
)

btc.head()
```

## 1. Binance BTC — Spot vs Perpetual

```python
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=btc["timestamp"],
        y=btc["binance_spot_mid"],
        mode="lines",
        name="Binance BTC Spot"
    )
)

fig.add_trace(
    go.Scatter(
        x=btc["timestamp"],
        y=btc["binance_perp_mid"],
        mode="lines",
        name="Binance BTC Perpetual"
    )
)

fig.update_layout(
    title="Binance BTC — Spot vs Perpetual",
    width=1200,
    height=600,
    xaxis_title="Время",
    yaxis_title="Цена BTC",
    hovermode="x unified"
)

fig.show()
```

## 2. Binance ETH — Spot vs Perpetual

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=eth["timestamp"],
        y=eth["binance_spot_mid"],
        mode="lines",
        name="Binance ETH Spot"
    )
)

fig.add_trace(
    go.Scatter(
        x=eth["timestamp"],
        y=eth["binance_perp_mid"],
        mode="lines",
        name="Binance ETH Perpetual"
    )
)

fig.update_layout(
    title="Binance ETH — Spot vs Perpetual",
    width=1200,
    height=600,
    xaxis_title="Время",
    yaxis_title="Цена ETH",
    hovermode="x unified"
)

fig.show()
```

## 3. SPB BTC — Perpetual vs Index

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=btc["timestamp"],
        y=btc["spb_perp_mid"],
        mode="lines",
        name="SPB BTC Perpetual"
    )
)

fig.add_trace(
    go.Scatter(
        x=btc["timestamp"],
        y=btc["spb_index_price"],
        mode="lines",
        name="SPB BTC Index"
    )
)

fig.update_layout(
    title="SPB BTC — Perpetual vs Index",
    width=1200,
    height=600,
    xaxis_title="Время",
    yaxis_title="Цена BTC",
    hovermode="x unified"
)

fig.show()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=eth["timestamp"],
        y=eth["spb_perp_mid"],
        mode="lines",
        name="SPB ETH Perpetual"
    )
)

fig.add_trace(
    go.Scatter(
        x=eth["timestamp"],
        y=eth["spb_index_price"],
        mode="lines",
        name="SPB ETH Index"
    )
)

fig.update_layout(
    title="SPB ETH — Perpetual vs Index",
    width=1200,
    height=600,
    xaxis_title="Время",
    yaxis_title="Цена ETH",
    hovermode="x unified"
)

fig.show()
```
